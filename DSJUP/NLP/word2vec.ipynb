```python
from nltk.corpus import stopwords 
import re
```

```python
def cleanText(text):
  text= text.lower()

  # parenthesis
  text= re.sub(r'\([^)]*\)', '', text)
  # \( -- opening parenthesis
  # \) -- closing parenthesis
  # [^)]*  - ^ except this(jehra v essto baad character likhya hoyyega ohnu shddke)  -- * zero or more

  # white multi-white space with single space
  text=re.sub(r'\s+',' ',text)
  # r'' raw text
  # \s any white space character +(one or more)


  text= re.sub(r'[“”‘’]', '', text)

  text=text.strip()

  return text
```

```python
data.loc[:,'New_text'] = data.loc[:,'text'].progress_apply(cleanText)
```

```python
from nltk.tokenize import word_tokenize
from nltk.stem import WordNetLemmatizer 
```

```python
nltk.download('punkt_tab')
nltk.download('wordnet')
```

```python
lemma = WordNetLemmatizer()
```

```python
l = ['hi','hello','hey','kaise ho','theek ho']
' '.join(l)
```

```python
def lemmatizeText(text):
    words = word_tokenize(text)
    lemmatize_words = [lemma.lemmatize(word) for word in words]
    return ' '.join(lemmatize_words)
```

```python
data.loc[:,'New_text'] = data.loc[:,'New_text'].progress_apply(lemmatizeText)
```

```python
from gensim.models import Word2Vec
model = Word2Vec(sentences = data.loc[:,'New_text'], window =3, vector_size = 10) 
# by default vector_size = 100, window = 5
```

```python
def convertVec(text):
    vectors = []
    for sentence in text:
        for word in sentence:
            if word in model.wv:
                vectors.append(model.wv[word])
    return vectors 
```
