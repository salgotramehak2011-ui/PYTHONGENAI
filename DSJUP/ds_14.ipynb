```python
import pandas as pd 
import numpy as np
```

```python
# map, applymap, getdummies

df = pd.DataFrame({'Name':['Sourav','Komal','Jatin','Anshu','Anurag','Simran','Reema'],
                  'Class':['btech','bca','mca','btech','mcom','mba','bba'],
                  'Rollno':[34,43,11,10,67,89,14]})
df
```

```python
df.loc[:,'Class'].str.capitalize()
```

```python
df.loc[:,'Rollno'] = df.loc[:,'Rollno']+2
df
```

```python
def funcCap(text):
    return text.capitalize()
```

```python
df.loc[:,'Class'] =  df.loc[:,'Class'].apply(funcCap) #series
df
```

```python
df.map(lambda x: str(x) + ' hi')
```

```python
df.applymap(lambda x: str(x) + ' hi')
```

```python
titanic = pd.read_csv('titanic-dataset.csv')
titanic
```

```python
titanic.Survived.value_counts()
```

```python
titanic.Survived = titanic.Survived.map({0:'Not Survived', 1:'Survived'})
titanic
```

```python
pd.get_dummies(titanic.Pclass,dtype = 'int')
```

```python
titanic.loc[:,[1,2,3]] = pd.get_dummies(titanic.Pclass, dtype='int')
titanic
```

```python
titanic.rename(columns={1:'Pclass1',2:'Pclass',3:'Pclass'},inplace=True)
titanic
```
